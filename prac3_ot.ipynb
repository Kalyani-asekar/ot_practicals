```python
import sympy as sp

x, y = sp.symbols('x y')

f = x**2 + y**2
constraint = x + y - 10

y_value = sp.solve(constraint, y)[0]

f_substituted = f.subs(y, y_value)

critical_x = sp.solve(sp.diff(f_substituted, x), x)

for x_value in critical_x:
    y_value = y_value
    print("Optimal point:", (x_value, y_value.subs(x, x_value)))
    print("Function value:", f_substituted.subs(x, x_value))
```
