```python
import sympy as sp

# Define the variable
x = sp.symbols('x')

# Define the function
f = x**3 - 6*x**2 + 9*x + 1

# Find the first derivative
f_prime = sp.diff(f, x)

# Find critical points by solving f'(x) = 0
critical_points = sp.solve(f_prime, x)

print("Critical points:", critical_points)

# Find the second derivative
f_double_prime = sp.diff(f_prime, x)

# Check each critical point
for point in critical_points:
    second_derivative = f_double_prime.subs(x, point)
    function_value = f.subs(x, point)

    if second_derivative > 0:
        print("Local Minimum at:", (point, function_value))
    elif second_derivative < 0:
        print("Local Maximum at:", (point, function_value))
    else:
        print("Second derivative test is inconclusive at:", point)
```
