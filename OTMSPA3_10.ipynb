```python
def floyd_warshall(graph):
    v = len(graph)
    dist = [list(row) for row in graph]
    for k in range(v):
        for i in range(v):
            for j in range(v):
                dist[i][j] = min(dist[i][j], dist[i][k] + dist[k][j])
    return dist

inf = float('inf')
graph = [
    [0, 5, inf, 10],
    [inf, 0, 3, inf],
    [inf, inf, 0, 1],
    [inf, inf, inf, 0]
]

shortest_paths = floyd_warshall(graph)

for row in shortest_paths:
    print([x if x != inf else "INF" for x in row])
```
