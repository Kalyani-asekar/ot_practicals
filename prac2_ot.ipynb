```python
import sympy as sp

x, y = sp.symbols('x y')

f = x**2 + y**2 - 4*x - 6*y

fx = sp.diff(f, x)
fy = sp.diff(f, y)

stationary_points = sp.solve([fx, fy], [x, y])

print("Stationary points:", stationary_points)
```
