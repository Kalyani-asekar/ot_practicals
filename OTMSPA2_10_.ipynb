```python
import sympy as sp

x, y, l = sp.symbols('x y l')

f = x**2 + y**2
g = x + y - 10

L = f - l * g

grad_L = [sp.diff(L, var) for var in (x, y, l)]

solutions = sp.solve(grad_L, (x, y, l), dict=True)

for sol in solutions:
    opt_val = f.subs(sol)
    print(f"Point: (x = {sol[x]}, y = {sol[y]}), Lambda = {sol[l]}, Objective Value = {opt_val}")
```
